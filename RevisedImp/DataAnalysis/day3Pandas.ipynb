## Merging and Combining DataFrames

When combining DataFrames, different join types determine how rows are matched and what data is included in the result.

*   **Inner Join (Intersection)**:
    *   Common rows only.

*   **Left Join (Difference in Set A.differnce(B), only A get but non-common in both and only B ele)**:
    *   All left rows, matching right rows.

*   **Right Join (Difference in SetB.differnce(A), only B get but non-common in both and only B ele)**:
    *   All right rows, matching left rows.

*   **Outer Join (Full Outer Join / Union)**:
    *   All rows from both, matching where possible.

```python
#merge() & join()
df = pd.merge(df1, df2, on="ID", how="inner") #by default inner
df
```

```python
pd.merge(df2, df1, on="ID", how="left")
```

**JOIN**

```python
#here in join we have to make any col as labell b'cuz this takes labell not index
df3 = df1.set_index("ID") #this have to do before doing join
df4 = df2.set_index("ID")
df3
df4
```

```python
df4.join(df3) #by default left join perform
```

```python
df3.join(df4, how="right")
```

```python
df3.join(df4, how="outer")
```
