```python
print(df.shape)      # Get number of rows & columns
print(df.info())     # Summary of DataFrame
print(df.describe()) # Statistical summary of numeric columns
print(df.columns)    # List of column names
print(df.dtypes)     # Data types of each column
```

**Modifying a DataFrame**

```python
#1.Adding a New Column
df["Salary"] = [50000, 60000, 70000]
df
```

```python
#2.update the col
df["Salary"] = df["Salary"] + 5000 #increase all salary by 5k
```

```python
df
```

```python
#3rd Drop col
df = df.drop(columns=["num3"])
```

```python
#4th Renaming Columns
df = df.rename(columns={"num1":"numbersOne"})
```

```python
df
```

**Sorting & Filtering Data**

```python
df = df.sort_values("Salary", ascending=False)
df
```

```python
df = df.sort_values("Salary", ascending=True)
df
```

```python
#Filtering the Rows
filtered_rows = df[df["Salary"] > 60000]
filtered_rows
```
