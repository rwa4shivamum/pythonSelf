# Mathematical Operations

```python
a1 = np.array([10,11])
a2 = np.array([5,14])
print(a1 + a2) #sum of both array
print(a1 * a2) #multiply of both array
print(a1 / a2) #division of both array
print(a1 - a2) #subtract of bith array
print(np.sqrt(a1)) #sqrt of a1 array
print(np.square(a1)) #sqaure of a1 array
ans = a1 ** 2
print(ans)
print(np.floor(a1))
print(np.ceil(a1))
print(np.round(a1))
print(np.concatenate((a1,a2))) #combine two array
```
