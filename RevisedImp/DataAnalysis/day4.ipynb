# Modifying a DataFrame

```python

df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "City": ["New York", "London", "Tokyo"]
})

df["Salary"] = [50000, 60000, 70000] #Adding extra Column
df["Age"] = df["Age"] + 1 #updating Age
df = df.drop(columns=["City"]) #dropping the column
df = df.rename(columns={"Name":"Full Name"}) #Renaming the column

df
```

# Sorting & Filtering Data

```python
df = df.sort_values("Age", ascending=False)
print(df)
```

```python
filtered_df = df[df["Age"] > 28]
filtered_df
```

# Handling Missing Data

```python
print(df.isnull().sum()) #count missing Values in Each Column
```

```python
df.fillna("Unknown", inplace=True) # Replace NaN with 'Unkown'
```

```python
df.dropna(inplace=True)#Drop row with missing value
df
```

# Aggregation & Grouping

```python
print(df["Age"].mean()) #Average age
```

```python
print(df["Age"].sum()) #Sum of all ages
```

```python
grouped = df.groupby("Full Name")["Age"].mean()
grouped
```

# Merging, Joining, and Concatenation

```python
df1 = pd.DataFrame({"ID": [1,2,3], "Name": ["A", "B", "C"]})
df2 = pd.DataFrame({"ID": [1,2,3], "Salary": [1000, 2000, 3000]})

merged_df = pd.merge(df1, df2, on="ID")
print(merged_df)
```

```python
df1 = pd.DataFrame({"A": [1, 2]})
df2 = pd.DataFrame({"A": [3, 4]})

concatenated = pd.concat([df1, df2])
print(concatenated)
```
