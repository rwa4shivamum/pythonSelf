```python
def withdraw(balance, amount):
    try:
        if amount > balance:
            raise ValueError("Insufficient balance")
        if amount < 0:
            raise ValueError("Invalid Amount")
        balance -= amount
        return "success"
    except:
        return      
```

```python
import matplotlib.pyplot as plt

# Sample Data
math_scores =    [45, 78, 56, 89, 90, 67, 80, 56, 92, 45,
                  60, 75, 85, 55, 70, 90, 88, 72, 60, 65,
                  68, 77, 85, 95, 48, 52, 60, 75, 80, 82]

science_scores = [40, 70, 60, 85, 88, 66, 75, 50, 90, 42,
                  58, 72, 80, 50, 65, 88, 86, 70, 55, 60,
                  63, 74, 83, 91, 45, 51, 58, 73, 78, 79]

plt.figure(figsize=(10,6))
plt.hist(math_scores, bins=7, color='blue', alpha=0.5, label='Math', edgecolor='black')
plt.hist(science_scores, bins=7, color='green', alpha=0.5, label='Science', edgecolor='black')
# Title and Labels
plt.title('Score Distribution: Math vs Science', fontsize=16)
plt.xlabel('Score Range', fontsize=12)
plt.ylabel('Number of Students', fontsize=12)

# Legend and Grid
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt

# Sample dataset
products = ['Laptop', 'Smartphone', 'Headphones', 'Monitor', 'Keyboard']
sales = [1200, 950, 300, 450, 200]

# Create a vertical bar chart (Column chart)
plt.figure(figsize=(8,5))
plt.bar(products, sales, color='skyblue', edgecolor='black')

# Add title and labels
plt.title('Product-wise Sales', fontsize=14)
plt.xlabel('Product', fontsize=12)
plt.ylabel('Sales ($)', fontsize=12)

# Optional: add value labels on top of bars
for i, v in enumerate(sales):
    plt.text(i, v + 20, str(v), ha='center', fontweight='bold')

# Show plot
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
ad_spend = [5, 6, 7, 6, 8, 7.5, 9, 10, 9.5, 11, 10.5, 12]
sales = [40, 45, 50, 46, 54, 53, 58, 62, 59, 65, 64, 70]
months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']

# Create Scatter Plot
plt.figure(figsize=(10,6))
plt.scatter(ad_spend, sales, color='darkorange', edgecolors='black', s=100)

# Annotate points with month names
for i in range(len(months)):
    plt.text(ad_spend[i]+0.1, sales[i], months[i], fontsize=9)

# Labels & Title
plt.title("Ad Spend vs Sales", fontsize=16)
plt.xlabel("Advertisement Spend ($1000s)", fontsize=12)
plt.ylabel("Sales ($1000s)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Dataset
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
revenue = [5000, 5200, 5800, 6200, 6000, 6700, 7100, 6900, 7200, 7500, 7700, 8000]

# Create a line chart
plt.figure(figsize=(10,6))
plt.plot(months, revenue, marker='o', linestyle='-', color='royalblue', linewidth=2)

# Add titles and labels
plt.title('Monthly Revenue in 2024', fontsize=16)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Revenue ($)', fontsize=12)

# Show value labels on data points
for i, value in enumerate(revenue):
    plt.text(i, value + 100, str(value), ha='center', fontsize=9)

# Add grid
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
