# Splitting Array

```python
import numpy as np

arr = np.array([1,2,3,4,5,6])
arr2 = np.split(arr,3)
print(arr2)#it equally split arr into proper 4 equal part in 2D array
```

```python
arr2D = np.array([[1,2,3,4],
                  [5,6,7,8]])
np.hsplit(arr2D, 2)
```

```python
np.vsplit(arr2D, 2)
```

# Sorting of Array

```python
a1 = np.array([4,3,2,1])
np.sort(a1)
```

```python
a2d = np.array([[3, 2, 1],
                [6, 5, 2],
                [7, 8, 2]])
a2d
```

```python
np.sort(a2d)#it short every single row in asc
```

```python
a2d[::-1]
```

```python
# find how to sort in descending order
np.sort(a1)[::-1]#1d array sort

#2D array sort
np.sort(a2d)[::-1][:,::-1] #reverse 2D array
#rvrs_arr[:,::-1] #reverse and sort each row in desc

# find how to sort col-wise & row-wise
np.sort(a2d)[::-1,:]#col wise sort in desc
np.sort(a2d)[:,::1]#col wise sort in asc

np.sort(a2d)#row-wise sort in asc
np.sort(a2d)[:,::-1]#row-wise sort in desc
#           [col,row]
```

```python
np.argsort(a2d)
```

```python
a2d = np.array([[3, 2, 1],
                [6, 5, 2],
                [7, 8, 2]])
np.sum(a2d, axis=1)#row sum
```

```python
np.sum(a2d,axis=0)#col sum
```

```python
#check if all aggregation & statistical function have "axis"
print(np.mean(a2d, axis=1))#mean of each row
print(np.mean(a2d, axis=0))#mean of each col
print(np.var(a2d, axis=1))#var of each row
print(np.std(a2d,axis=1))#std of each row
print(np.median(a2d, axis=1))#medain of each row
# parameter to work with 2d array ?
# help(np.std())
```
