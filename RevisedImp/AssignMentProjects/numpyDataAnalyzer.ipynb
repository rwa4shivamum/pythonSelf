# Numpy Analyzer

```python
import numpy as np
class DataAnlytics():
    def __init__(self):
        self.array = None
        
    #_____Private Method______
    def _validateShape_Size(self,arr2):
        return np.size(self.array) == np.size(arr2) and np.shape(self.array) == np.shape(arr2)
    #____Static Method_____
    @staticmethod
    def show_menu():
        print("""
        ===== NumPy Analyzer =====
        1. Create Array
        2. Math Operations
        3. Search / Sort / Filter
        4. Aggregations
        5. Exit
        """)
        
    def create_array(self):
        print('''Select type of Array to Create:
              1.1D array
              2.2D array
              3.3D array
              ''')
        choice = int(input("Enter your choice: "))
        #1D Array
        if choice == 1:
            self.array = np.array(list(map(lambda a: int(a),input("Enter a numb").split(" "))))
            
        #2D Array
        elif choice == 2:
            row = int(input("Enter Row"))
            col = int(input("Enter col"))
            enterElemt = row * col
            
            lst = list(map(int, input(f"Enter value:{enterElemt} ").split(" ")))
            self.array = np.array(lst).reshape(row, col)
        
        #3D array
        elif choice == 3:
            noOf2Ddep = int(input("Enter no. of depth: "))
            noOfRow = int(input("Enter no. of rows: "))
            noOfCol = int(input("Enter no. of cols: "))
            expectValue = noOfRow * noOfCol
            
            matrix3D = []
            
            for i in range(noOf2Ddep):
                lst = list(map(int, input(f"Enter value:depth{i} {expectValue}").split(" ")))
                arr2d = np.array(lst).reshape(noOfRow,noOfCol)
                matrix3D.append(arr2d)
            
            self.array = np.array(matrix3D)
        else:
            print("\nInvalid Choice")
        
        print("\nCreated Array:\n", self.array)
    
    def get_array(self):
        return self.array
    
    def math_operations(self,arr2):
        
        if(self._validateShape_Size(arr2)):
            print('''
                  Choose Mathematical Operation
                  1.Addition 
                  2.Subtraction
                  3.multiplication
                  4.Division
                  5.Dot Product
                  6.Matrix Multiplication
                  ''')
            choiceOfOperation = int(input("Enter a operation: "))
            if choiceOfOperation == 1:
                return self.array + arr2
            elif choiceOfOperation == 2:
                return self.array - arr2
            elif choiceOfOperation == 3:
                return self.array * arr2
            elif choiceOfOperation == 4:
                return self.array / arr2
            elif choiceOfOperation == 5:
                return np.dot(self.array,arr2)
            elif choiceOfOperation == 6:
                return np.matmul(self.array, arr2)
            else:
                return "Invalid"
        else:
            return "Shape And Size was not Same"
            
    def search_sort_filter(self):
        print('''
              Choice of operation
              1.Search
              2.Sort
              3.Filter
              ''')
        choiceOfSearchSortFil = int(input("Enter a choice: "))
        if choiceOfSearchSortFil == 1:
            valueSearch = int(input("Enter a value: "))
            return np.where(self.array == valueSearch)
        elif choiceOfSearchSortFil == 2:
            return np.sort(self.array)
        elif choiceOfSearchSortFil == 3:
            enterValueFil = int(input("Enter the number: "))
            return self.array[self.array > enterValueFil]
        else:
            return "Invalid Input"
    
    def aggregations(self):
        return {
            "min": np.min(self.array),
            "max": np.max(self.array),
            "mean": np.mean(self.array),
            "sum": np.sum(self.array),
            "variance": np.var(self.array),
            "std_dev": np.std(self.array),
            "median": np.median(self.array)
        }
        
def main():
    obj = DataAnlytics()

    while True:
        DataAnlytics.show_menu()
        choice = int(input("Enter choice: "))

        if choice == 1:
            obj.create_array()

        elif choice == 2:
            print("Create second array:")
            obj2 = DataAnlytics()
            obj2.create_array()
            result = obj.math_operations(obj2.array)
            print("Result:\n", result)

        elif choice == 3:
            result = obj.search_sort_filter()
            print("Result:\n", result)

        elif choice == 4:
            result = obj.aggregations()
            for k, v in result.items():
                print(f"{k}: {v}")

        elif choice == 5:
            print("Exiting...")
            break

        else:
            print("Invalid choice")

main() 
```
