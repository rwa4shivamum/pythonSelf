# Boolean Mask (Masking Array)

```python
arr1 = np.array([10,12,13,14,15])
m = arr1%2==0 #this get true if num is disible by 2
m
```

```python
np.ma.array(arr1,mask=m)#from here falsy get and truthy become --
```

```python
odd_number = np.ma.array(arr1, mask=arr1%2!=0)
print(odd_number)
```
