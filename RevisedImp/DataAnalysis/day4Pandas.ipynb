**Re_Indexing & Altering Labels**

```python
data = {
    'Name':['Alice', 'Bob', 'Charlie'],
    'Age':[25, 30, 35],
}
data
```

```python
df = pd.DataFrame(data,index=['a', 'b', 'c'])
df
```

```python
df.reindex(['c', 'a', 'b'])#this show the data in our ouder wise by label
```

```python
df.reindex(['c','a','b','p'])
#this makes the extra index
```

```python
df = df.reindex(columns=['Age', 'Name'])
df
#this makes the column
#we can also rearrage the column
#if i make inplace=True
```

```python
df
```

```python
#Rename index and columns
df.rename(index={'a':'b'}, columns={'Name':'First Name'}, inplace=True)
df
```

```python
#Directly Replace Index or column Labels
df.index = ['X','Y','Z']
df.columns = ['PersonName','PersonAge']
df
```

**Pivoting** :summary

```python
# To perform pivoting in python
# i.pivot
# ii.pivot_table(add Aggregation func)
data = {
    "Date":["2024-01-01", "2024-01-01", "2024-01-02", "2024-01-02"],
    "Product":["Laptop", "Phone", "Laptop", "Phone"],
    "Sales":[1000, 500, 1500, 800]
}

df = pd.DataFrame(data)
df
```

```python
df.pivot(index='Date', columns="Product", values="Sales")
#we get product in Column and
```
